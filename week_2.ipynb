Exception Handling and Debugging in Python

```python
try:
    x = int(input("Enter a number: "))
    print("You entered:", x)
except ValueError:
    print("That was not a valid number!")
```

```python
try:
    num = int(input("Enter a number to divide 100: "))
    result = 100 / num
except ZeroDivisionError:
    print("Cannot divide by zero!")
except ValueError:
    print("Invalid input! Not a number.")
else:
    print("Result is:", result)
finally:
    print("Execution complete. This block always runs.")
```

```python
try:
    a = int("ten")  # Causes ValueError
    b = 10 / 0      # Causes ZeroDivisionError
except (ValueError, ZeroDivisionError) as e:
    print("An error occurred:", e)
```

```python
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative!")
    else:
        print("Age is valid.")

try:
    check_age(-5)
except ValueError as e:
    print("Caught an exception:", e)
```

```python
def divide(a, b):
    print(f"Dividing {a} by {b}")  # Debug info
    return a / b

print(divide(10, 2))
```

```python
import traceback

try:
    result = 10 / 0
except Exception as e:
    print("Something went wrong.")
    traceback.print_exc()
```

collections – Specialized Containers
🔹 A. Counter

```python
from collections import Counter
fruits = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']
count = Counter(fruits)
print("Fruit Count:", count)
```

 B. defaultdict

```python
from collections import defaultdict
marks = defaultdict(lambda: 0)
marks['Math'] = 90
marks['Science'] = 80
print("English Marks (not set):", marks['English'])
```

🔹 C. namedtuple


```python
from collections import namedtuple
Point = namedtuple('Point', 'x y')
p1 = Point(10, 20)
print("Point:", p1)
print("x:", p1.x, "y:", p1.y)
```

D. deque

```python
from collections import deque
dq = deque([1, 2, 3])
dq.append(4)
dq.appendleft(0)
dq.pop()
dq.popleft()
print("Deque after operations:", dq)
```

1.1 Generate a list of 100 simulated sensor readings.
  Each reading should be a tuple: (timestamp_str, device_id, device_type, location, value)
  - timestamp_str: Format as 'YYYY-MM-DD HH:MM:SS'
   - device_id: A unique ID for each reading (e.g., 'DEV001', 'DEV002', etc., up to 100 unique IDs).
   - device_type: Randomly chosen from DEVICE_TYPES.
    - location: Randomly chosen from LOCATIONS.
  - value: Random float within READING_RANGES for its device_type (round to 2 decimal places).
 Store these in a list called 'raw_sensor_data'.

Your code for Task 1.1 here:

```python
import random
from datetime import datetime

DEVICE_TYPES = ["Temperature", "Humidity", "Pressure", "light"]

LOCATIONS = ["Lab 1", "lab 2", "officeA", "officeB", "warehouse"]

READING_RANGES = {
    "Temperature": (18.0, 30.0),
    "Humidity": (30.0, 80.0),
    "Pressure": (980.0, 1030.0),
    "light": (100, 1000)
}

raw_sensor_data = []

for i in range(1, 101):
    device_type = random.choice(DEVICE_TYPES)
    low, high = READING_RANGES[device_type]

    raw_sensor_data.append((
        datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        f"DEV{i:03d}",
        device_type,
        random.choice(LOCATIONS),
        round(random.uniform(low, high), 2)
    ))

for reading in raw_sensor_data:
    print(reading)
```

TASK 1.2 From 'raw_sensor_data', create a list named 'temperature_readings'     containing only the readings
     where 'device_type' is 'Temperature_Sensor'.
     Each item in 'temperature_readings' should be a dictionary with keys:
     'timestamp', 'device_id', 'location', 'temperature_celsius'.

Your code for Task 1.2 here:


```python
temperature_readings = []

for reading in raw_sensor_data:
    if reading[2] == "Temperature":
        temperature_readings.append({
            "timestamp": reading[0],
            "device_id": reading[1],
            "location": reading[3],
            "temperature_celsius": reading[4]
        })

for item in temperature_readings[:5]:
    print(item)
```

TASK 1.4 Create a dictionary named 'device_stats' where keys are 'device_type' and values are a tuple: (min_value, max_value) observed for that device type across all readings Initialize with sensible large/small values.
Your code for Task 1.4 here:

```python
device_stats = {}

for reading in raw_sensor_data:
    device_type = reading[2]
    value = reading[4]

    if device_type not in device_stats:
        device_stats[device_type] = (float('inf'), float('-inf'))

    min_val, max_val = device_stats[device_type]

    device_stats[device_type] = (
        min(min_val, value),
        max(max_val, value)
    )

print(device_stats)
```
